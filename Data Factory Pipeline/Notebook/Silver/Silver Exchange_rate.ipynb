```python
# Đọc dữ liệu exchange_rate từ Bronze

from pyspark.sql import functions as F

bronze_exchange_rate_path = "Files/Bronze/exchange_rate"

df_exchange_rate = spark.read.parquet(bronze_exchange_rate_path)

display(df_exchange_rate)
```

```python
# Ghi dữ liệu exchange_rate vào tầng Silver

df_exchange_rate.write\
    .format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .saveAsTable("silver_exchange_rate")
```

```python
display(spark.sql(
    "SELECT *\
    FROM silver_exchange_rate"
))
```
